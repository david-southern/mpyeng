```python
import cadquery as cq
from jupyter_cadquery import (
    show, open_viewer, close_viewers, set_defaults,
    PartGroup, Part
)
from jupyter_cadquery.export import exportSTL

set_defaults(
    axes=True, 
    grid=(True, True, True),
    ortho=False,
    tree_width=250,
    cad_width=640,
    default_color=(232, 196, 64),
    default_edge_color="#E800D0",
#    js_debug=True
)

close_viewers()
cv = open_viewer("CadQuery", anchor="right", height=800)
```

```python
exportObject = False

box = cq.Workplane('XY').box(10, 1, 1).edges().fillet(0.3)
show(box, viewer="CadQuery")

box0 = cq.Workplane('XY').box(10, 20, 30).edges(">X or <X").chamfer(2)
box0.name = "box1"

box2 = cq.Workplane('XY').box(8, 18, 28).edges(">X or <X").chamfer(2)
box2.name = "box2"

box3 = cq.Workplane('XY').transformed(offset=(0, 15, 7))\
    .box(30, 20, 6).edges(">Z").fillet(3)
box3.name = "box3"

box4 = box3.mirror("XY").translate((0, -5, 0))
box4.name = "box4"

box1 = box0\
    .cut(box2)\
    .cut(box3)\
    .cut(box4)

if(exportObject):
    exportSTL(Part(box1), "boxer.stl", tolerance=0.1, angular_tolerance=0.1)
```

```python
box0 = cq.Workplane('XY').box(10, 20, 30).edges(">X or <X").chamfer(2)
box0.name = "box1"

box2 = cq.Workplane('XY').box(8, 18, 28).edges(">X or <X").chamfer(2)
box2.name = "box2"

box3 = cq.Workplane('XY').transformed(offset=(0, 15, 7))\
    .box(30, 20, 6).edges(">Z").fillet(3)
box3.name = "box3"

box4 = box3.mirror("XY").translate((0, -5, 0))
box4.name = "box4"

box1 = box0\
    .cut(box2)\
    .cut(box3)\
    .cut(box4)

show(box1, viewer="CadQuery")
```

```python
a1 = PartGroup(
    [
        Part(box1, "red box",   "#d7191c", show_edges=False),
        Part(box3, "green box", "#abdda4", show_edges=True),
        Part(box4, "blue box",  "#2b83ba", show_faces=False),
    ],
    "example 1"
)

cv = show(a1, viewer="CadQuery", \
          axes=True, grid=[True, False, False], \
          axes0=True)
```

```python
import cadquery as cq
from jupyter_cadquery.export import exportSTL
from jupyter_cadquery import Part, PartGroup, show

box = cq.Workplane().box(10,20,30).edges().fillet(4)
sphere = cq.Workplane().sphere(12).translate((0,0,10))

quality = 0.1
angular_tolerance = 0.1

pg = PartGroup([Part(box),Part(sphere)])

exportSTL(
    pg,
    "thingy.stl", 
    tolerance=quality, 
    angular_tolerance=angular_tolerance
)
```
